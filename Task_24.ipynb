```python
import pandas as pd

df = pd.read_csv("SampleSuperstore.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Data Quality Audit

audit_results = []

# 1. Missing values
for column in df.columns:
    missing_count = df[column].isnull().sum()
    audit_results.append({
        "Check": "Missing Values",
        "Column": column,
        "Issue Count": missing_count
    })

# 2. Duplicate rows
duplicate_count = df.duplicated().sum()
audit_results.append({
    "Check": "Duplicate Rows",
    "Column": "All Columns",
    "Issue Count": duplicate_count
})

# 3. Invalid Quantity
invalid_quantity = (df["Quantity"] <= 0).sum()
audit_results.append({
    "Check": "Invalid Quantity",
    "Column": "Quantity",
    "Issue Count": invalid_quantity
})

# 4. Invalid Sales
invalid_sales = (df["Sales"] < 0).sum()
audit_results.append({
    "Check": "Invalid Sales",
    "Column": "Sales",
    "Issue Count": invalid_sales
})

# 5. Invalid Discount
invalid_discount = ((df["Discount"] < 0) | (df["Discount"] > 1)).sum()
audit_results.append({
    "Check": "Invalid Discount",
    "Column": "Discount",
    "Issue Count": invalid_discount
})

# 6. Invalid Postal Code
invalid_postal = (df["Postal Code"] <= 0).sum()
audit_results.append({
    "Check": "Invalid Postal Code",
    "Column": "Postal Code",
    "Issue Count": invalid_postal
})

# Convert results to DataFrame
audit_df = pd.DataFrame(audit_results)

audit_df
```

```python
# Consistency Checks

print("Ship Mode values:")
print(df["Ship Mode"].unique())

print("\nSegment values:")
print(df["Segment"].unique())

print("\nCountry values:")
print(df["Country"].unique())

print("\nRegion values:")
print(df["Region"].unique())

print("\nCategory values:")
print(df["Category"].unique())

print("\nSub-Category values:")
print(df["Sub-Category"].unique())
```

```python
# Create Issue Log

issue_log = pd.DataFrame({
    "Issue Type": [
        "Duplicate Rows"
    ],
    "Column": [
        "All Columns"
    ],
    "Issue Count": [
        17
    ],
    "Severity": [
        "Medium"
    ],
    "Action Taken": [
        "Duplicate rows removed"
    ]
})

issue_log
```

```python
issue_log.to_csv("Issue_Log.csv", index=False)

print("Issue Log saved successfully!")
```

```python
# Remove duplicate rows

cleaned_df = df.drop_duplicates().copy()

print("Original rows:", len(df))
print("Rows after removing duplicates:", len(cleaned_df))
print("Rows removed:", len(df) - len(cleaned_df))
```

```python
# Final Quality Check

print("Final Dataset Shape:", cleaned_df.shape)

print("\nMissing Values:")
print(cleaned_df.isnull().sum())

print("\nDuplicate Rows:", cleaned_df.duplicated().sum())

print("\nInvalid Quantity:", (cleaned_df["Quantity"] <= 0).sum())

print("\nInvalid Sales:", (cleaned_df["Sales"] < 0).sum())

print("\nInvalid Discount:",
      ((cleaned_df["Discount"] < 0) | (cleaned_df["Discount"] > 1)).sum())

print("\nInvalid Postal Code:",
      (cleaned_df["Postal Code"] <= 0).sum())
```
